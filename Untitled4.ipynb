```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as ply
```

```python
x= np.linspace(20,100,20)
y=np.sqrt(x)
z=np.log(x*x)

ply.plot(x,y,"r+",x,z,"g^")
ply.show()
```

```python
ply.subplot(121)
ply.plot(x,y,"r+")
ply.xlabel("Number_1")
ply.ylabel("Number 2")
ply.title("SQRT")

ply.subplot(122)
ply.plot(x,z,"g*")
ply.xlabel("Number_1")
ply.ylabel("Number 2")
ply.title("SQRT LOG")

ply.show()
```
