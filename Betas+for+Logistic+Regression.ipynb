```python
churn_data = pd.read_csv("churn_data.csv")
internet_data = pd.read_csv("internet_data.csv")
customer_data = pd.read_csv("customer_data.csv")
```

```python
#internet_data["OnlineBackup"].value_counts()

#internet_data["OnlineSecurity"].value_counts()
internet_data["InternetService"].value_counts()
```

```python
internet_data[internet_data.DeviceProtection =='No internet service']
```

```python
pd.set_option('display.max_columns', 500)
df = pd.merge(churn_data,internet_data,how="inner",on="customerID")
df = pd.merge(df,customer_data,how="inner",on="customerID")

df.head()
```

```python
ls_catfields =["PhoneService","PaperlessBilling","Churn","OnlineSecurity","OnlineBackup","DeviceProtection","TechSupport",
               "StreamingTV","StreamingMovies","Partner","Dependents"]

def binary_mapping(x):
    return x.map({"Yes":1,"No" :0})

df[ls_catfields] = df[ls_catfields].apply(binary_mapping)
```

```python
df.head()
```

```python
# dummy variables

lsMulticlass =["Contract","PaymentMethod","MultipleLines","InternetService","gender"]

dummies = pd.get_dummies(df[lsMulticlass],drop_first =True)

df =pd.concat([df,dummies],axis =1)

df.head()
```
