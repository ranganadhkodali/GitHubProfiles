```python
n = 5

chars = "abcdefghijklmnopqrstuvwxyz"
lst = [None] * ((n * 2) -1)

initial =chars[:n]
ss=initial[::-1]+initial[1:]

middel = "-".join([i for i in ss])
lst[n-1]  = middel



for i in range(1,n):
    up_indent = n+i-1
    low_indent = n-i-1
    initial = initial[1:]
    ss=initial[::-1]+initial[1:]
    
    lst[up_indent] =  "-"*(i*2) + "-".join([i for i in ss]) + "-"*(i*2)     
    lst[low_indent] = "-"*(i*2) + "-".join([i for i in ss]) + "-"*(i*2)

for i in lst:
    print(i)
```
