```python
import numpy as np
seed(10)
population =np.arange(1,1000)


print("Population Mean", round(population.mean(),2))

plt.title("Population Distribution")
plt.hist(population)
plt.show()

#sample mean for 10 samples
Samples =[]
for i in np.arange(1,5):
    item =np.random.choice(population,20)
    Samples.append(item)
    

sample_means = [mean(i) for i in Samples]

print("Means for each sample : ",sample_means)


plt.title("Sample Mean of 5 samples = " + str(round(mean(sample_means),2) ))
plt.hist(sample_means)
plt.show()

# exctra the 50 samples
Samples =[]
for i in np.arange(1,50):
    item =np.random.choice(population,20)
    Samples.append(item)
    

sample_means = [mean(i) for i in Samples]

plt.title("Sample Mean of 50 samples = " + str(round(mean(sample_means),2) ))
plt.hist(sample_means)
plt.show()
```

```python
from functools import reduce
import operator as op
def ncr(n, r):
    r = min(r, n-r)
    numer = reduce(op.mul, range(n, n-r, -1), 1)
    denom = reduce(op.mul, range(1, r+1), 1)
    return numer / denom

```
