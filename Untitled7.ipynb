```python
import collections

def flatten_dict(dd, separator='_', prefix=''):
    return { prefix + separator + k if prefix else k : v
             for kk, vv in dd.items()
             for k, v in flatten_dict(vv, separator, kk).items()
             } if isinstance(dd, dict) else { prefix : dd }
```

```python
import re
def validate(email): 
    email = email.lower()
    
    match=re.match('^[a-z0-9]+(\[a-z0-9]+)*@[a-z]+(\.[a-z])*(\.[a-z]{2,3})$',email)
    _list =email.split('.')
    
    if(len(_list) != 2):
        return 'Invalid email.'
    
    if (match == None) :
        return 'Invalid email.'
    else:
        return 'Valid email.'
```

```python
validate('abaaba@tt@.com')
```

```python
def emailValidate(email):
    if not re.match(r"-/\#%&*().!*$", email):
         return 'Invalid email.' 
    
    _list =email.split(r'@')
      
    if (len(_list) !=2):
        return 'Invalid email.' 
    match = re.search(r"[A-Z0-9a-z]",_list[0])
    if not match:
        return 'Invalid email.' 
    _list1 = _list[1].split(".")
    
    if (len(_list1) !=2):
        return 'Invalid email.' 
       
    match =re.search(r"[A-Za-z]",_list1[0])
       
    if not match:
        return 'Invalid email.' 
       
    match = re.search("[A-Za-z]{2,3}$",_list1[1])
    
    if not match:
        return 'Invalid email.' 
       
    return 'Valid email.' 
```

```python
string="ddddaacccb"
n=3
```

```python
_list = pd.DataFrame([i for i in string])

alpha ="abcdefghijklmnopqrstuvwxyz"

_list.columns =['alpha']
_list["val"] =1

_list1=_list.groupby("alpha").count()
_list1=_list1.reset_index(level=0)

#list(_list1[_list1.val <=n].sort_values(by="alpha",ascending =True).sort_values(by="val",ascending =False).alpha)

freq=list(set(_list1.sort_values("val",ascending =False).val))
#print(list(_list1[_list1.val <=n].sort_values("alpha").alpha))

freq=sorted(freq,reverse=True)

final=[]

for i in freq[:n]:
    sublist=_list1[_list1.val==i]
    sublist=sorted(sublist.alpha, key=str.lower)
    final.append(sublist[0])
    
    
print(sorted(final))
```

```python
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.svm import SVC

data = datasets.load_iris()
X = data.data
y = data.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)

model = SVC()
model.fit(X_train_scaled, y_train)

X_test_scaled = scaler.transform(X_test)
y_pred = model.predict(X_test_scaled)
```
