```python
import statsmodels
import statsmodels.api as sm 

import sklearn 
from sklearn.model_selection import train_test_split

data.head()
```

```python
X = data["TV"]
y= data["Sales"]

y_train,y_test,X_train,X_test=train_test_split(y,X,test_size=0.3,train_size=0.7,random_state=100)
```

```python
X_train_SM =sm.add_constant(X_train)
```

```python
sm_model = sm.OLS(y_train,X_train_SM)
fit = sm_model.fit()

fit.summary()
```

```python
fit.params.TV
```

```python
plt.plot(x=X_train,y=y_train)
plt.plot(x=X_train,y=fit.predict())


ax = sns.regplot(x=X, y=y, line_kws={'label':'$y=%3.7s*x+%3.7s$'%(fit.params.TV, fit.params.const)},fit_reg =False);
ax.legend()
sns.regplot(x=X_train, y=y_train, fit_reg=True, ax=ax,color="r",marker="*",x_jitter=.1);
sns.regplot(x=X_train, y=fit.predict(),scatter=False, ax=ax,x_jitter=.3);
plt.show()
```

```python
#plt.plot( y - pd.Series(fit.predict()) ,"ro")
res = y_train -  fit.predict(X_train_SM)

sns.distplot(res)
```

```python
sns.boxplot( (y - pd.Series(fit.predict())) )
```

```python
sns.scatterplot(x=X_train,y=res)
plt.xlabel("TV")
plt.ylabel("Residual")
plt.title("TV vs Residual")
plt.show()
```

```python
fit.predict(X_train_SM).describe()
```

```python
X_test_SM =sm.add_constant(X_test)

pred = fit.predict(X_test_SM)

residual = y_test - pred
```

```python
ax=sns.regplot(x=X_test,y=y_test,fit_reg=False)
sns.regplot(x=X_test,y=pred,fit_reg=True,color="r",ax=ax)
```

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
```

```python
r2 = r2_score(y_test,pred)

r2
```

```python
#help(mean_squared_error)
mean_squared_error(y_test,pred)
```

```python
sns.scatterplot(x=X_test,y=residual)
```

```python
votes =["pasta","past1a","pasta","pasta","pasta","paratha","paratha","paratha"]

dic ={}
for x in votes:
    if not x in  dic:      
        dic[x] = votes.count(x)
    
lead_item =sorted(dic.items(), key=lambda x: x[1],reverse =True)[0]

if (lead_item[1] >= (len(votes) /2+1)):
    print(lead_item[0])
else:
    print("NOTA")
```
