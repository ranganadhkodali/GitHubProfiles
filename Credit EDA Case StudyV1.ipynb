## Data types review 

a. Data types reasonable in the correct form. with the most categorical variables such as example FLAG_MOBIL,FLAG_EMP_PHONE,FLAG_WORK_PHONE,FLAG_CONT_MOBILE,FLAG_PHONE,FLAG_EMAIL are in the numaric form with 1 and 0 and 1 indicates client has a True Flag and 0 seems indicating client does't own or does't have specified object. it is better to convert the these categorical columns into the "Y" and "N" to make data set easly readable.

b. numaric columns DAYS_BIRTH,DAYS_EMPLOYED,DAYS_REGISTRATION,DAYS_ID_PUBLISH are in -ve it is better to convert these columns to +ve will be more readable.


```python
_numarical_cat_columns =['FLAG_MOBIL','FLAG_EMP_PHONE','FLAG_WORK_PHONE','FLAG_CONT_MOBILE','FLAG_PHONE','FLAG_EMAIL',
                         'REG_REGION_NOT_LIVE_REGION','REG_REGION_NOT_WORK_REGION','LIVE_REGION_NOT_WORK_REGION',
                        'REG_CITY_NOT_LIVE_CITY','REG_CITY_NOT_WORK_CITY','LIVE_CITY_NOT_WORK_CITY','FLAG_DOCUMENT_2','FLAG_DOCUMENT_3',
                         'FLAG_DOCUMENT_4','FLAG_DOCUMENT_5','FLAG_DOCUMENT_6','FLAG_DOCUMENT_7','FLAG_DOCUMENT_8','FLAG_DOCUMENT_9',
                         'FLAG_DOCUMENT_10','FLAG_DOCUMENT_11','FLAG_DOCUMENT_12','FLAG_DOCUMENT_13','FLAG_DOCUMENT_14','FLAG_DOCUMENT_15',
                         'FLAG_DOCUMENT_16','FLAG_DOCUMENT_17','FLAG_DOCUMENT_18','FLAG_DOCUMENT_19','FLAG_DOCUMENT_20','FLAG_DOCUMENT_21']

for i in _numarical_cat_columns:
    applications[i] =applications[i].astype(str)
    applications[i] = ["Y" if i=="0" else "N" for i  in  applications[i]]
```

```python
applications.head()
```

```python
# covert number of days into +ve 

applications["DAYS_BIRTH"]=applications["DAYS_BIRTH"] *-1
applications["DAYS_EMPLOYED"]=applications["DAYS_EMPLOYED"] *-1
applications["DAYS_REGISTRATION"]=applications["DAYS_REGISTRATION"] *-1
applications["DAYS_ID_PUBLISH"]=applications["DAYS_ID_PUBLISH"] *-1
```
