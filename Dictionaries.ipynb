```python
matrix_string="1 2\n10 20"

rows =[[int(cell) for cell in row.split(' ')] 
       for row in matrix_string.split('\n')]


columns = [list(x) for x in zip(*rows)]

print(rows)
print(columns)
```

```python
class Matrix(object):
    def __init__(self, matrix_string):
        self.rows =[[int(cell) for cell in row.split(' ')] \
                               for row in matrix_string.split('\n')]
        self.cols = [list(x) for x in zip(*rows)]


    def row(self, index):
        return self.rows[index-1]

    def column(self, index):
        return self.cols[index-1]

matrix = Matrix("1 2\n10 20")
matrix.row(1)
```
