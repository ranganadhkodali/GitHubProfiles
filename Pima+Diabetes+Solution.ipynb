```python
import numpy as np
import pandas as pd
import json
from sklearn.linear_model import LogisticRegression
from sklearn import datasets, linear_model, metrics

from sklearn import datasets, linear_model, metrics

# read training data
diabetes_train = pd.read_csv("/data/training/diabetes_train.csv")

# read test data
diabetes_test = pd.read_csv("/data/test/diabetes_test.csv")

# read the file containing actual labels (for computing scores)
actual_data = pd.read_csv("https://media-doselect.s3.amazonaws.com/generic/PZ2x3L59Q1k7jbRv02BJJjjZY/sample_diabetes_actual.csv")


print(diabetes_train.head())
print(diabetes_test.head())
print(actual_data.head())


## WRITE YOUR CODE HERE ##

# create logistic regression object
log_model = LogisticRegression()

#  create x_train and y_train
x_train = diabetes_train.iloc[:, :7]
y_train = diabetes_train.iloc[:, 7]
y_true = actual_data.iloc[:, 0]


# Fit the model
log_model.fit(x_train,y_train)

 

# make predictions
y_pred = log_model.predict(diabetes_test.iloc[:, :7])

 

accuracy_score = round(metrics.accuracy_score(y_true, y_pred),2)
precision_score = round(metrics.precision_score(y_true, y_pred),2)
recall_score = round(metrics.recall_score(y_true, y_pred),2)
f1_score = round(metrics.f1_score(y_true, y_pred),2)

 

print(accuracy_score, precision_score, recall_score, f1_score)


# write the scores into the output file as a dictionary
# this code is already written for you
d = {'recall_score': recall_score,
    'f1_score': f1_score,
    'accuracy_score': accuracy_score,
    'precision_score': precision_score}
print(d)

# writing the dictionary to output file
# code is already written
with open('/code/output/answers.txt', 'w') as file:
     file.write(json.dumps(d))

flag
0  upvotes  -  1  comments
thumb_upUpvote


pd.set_option('display.max_columns', 500)
```
