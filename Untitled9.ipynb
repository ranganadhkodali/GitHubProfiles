```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import datetime as dt

import sklearn
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

from scipy.cluster.hierarchy import linkage
from scipy.cluster.hierarchy import dendrogram
from scipy.cluster.hierarchy import cut_tree
```

```python
data = pd.read_csv("Cricket.csv",sep=",", encoding="ISO-8859-1", header=0)
```

```python
data.head()
```

```python
data1=data[["Ave","SR"]].copy()
```

```python
data1.info()
```

```python
scaler = StandardScaler ()

data2 = scaler.fit_transform(data1)
data2 =  pd.DataFrame(data2)
data2.columns = data1.columns 
data2.head()
```

```python
kmeans = KMeans(n_clusters=4, max_iter=50,random_state=100)
kmeans.fit(data2)
```

```python
data["cls"] = kmeans.labels_
```

```python
data.head()
```

```python
data[data.cls==3]
```

```python
from scipy.cluster.hierarchy import linkage
from scipy.cluster.hierarchy import dendrogram
from scipy.cluster.hierarchy import cut_tree
```

```python
mergings = linkage(data2, method="complete", metric='euclidean')
dendrogram(mergings)
plt.show()
```

```python
cluster_labels = cut_tree(mergings, n_clusters=4).reshape(-1, )
cluster_labels
```

```python
data["cls1"] =cluster_labels
```
